## Papraphraser

```python
from datasets import load_dataset

paraphraser = load_dataset("merionum/ru_paraphraser")
paraphraser
```

```python
paraphraser_train = paraphraser["train"].to_pandas()

paraphraser_test = paraphraser["test"].to_pandas()
```

```python
paraphraser_train.head()
```

```python
paraphraser_train["class"].value_counts().sort_index()
```

```python
paraphraser_train_bin = paraphraser_train[
    paraphraser_train["class"].isin(['-1', '1'])
].copy()

paraphraser_train_bin = paraphraser_train_bin[
    ["text_1", "text_2", "class"]
].rename(columns={
    "text_1": "sentence1",
    "text_2": "sentence2",
    "class": "label"
})

paraphraser_train_bin["label"] = paraphraser_train_bin["label"].astype(int)
paraphraser_train_bin["label"] = paraphraser_train_bin["label"].map({

    -1: 0,
    1: 1

})
```

```python
paraphraser_train_bin["label"].value_counts()
```

```python
paraphraser_test_bin = paraphraser_test[
    paraphraser_test["class"].isin(['-1', '1'])
].copy()

paraphraser_test_bin = paraphraser_test_bin[
    ["text_1", "text_2", "class"]
].rename(columns={
    "text_1": "sentence1",
    "text_2": "sentence2",
    "class": "label"
})

paraphraser_test_bin["label"] = paraphraser_test_bin["label"].astype(int)
paraphraser_test_bin["label"] = paraphraser_test_bin["label"].map({

    -1: 0,
    1: 1

})
```

```python
paraphraser_test_bin["label"].value_counts()
```

## RuPaws

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data")
RUPAWS_DIR = DATA_DIR / "rupaws_dataset"
```

```python
qqp_train = pd.read_csv(
    RUPAWS_DIR / "train/rupaws_qqp_train.csv",
    sep=";"
)

wiki_train = pd.read_csv(
    RUPAWS_DIR / "train/rupaws_wiki_train.csv",
    sep=";"
)

qqp_train.head()
```

```python
wiki_train.head()
```

```python
wiki_train = wiki_train[["sentence1", "sentence2", "label"]]
```

```python
rupaws_train = pd.concat(
    [qqp_train, wiki_train],
    ignore_index=True
)

rupaws_train.shape
```

```python
qqp_test = pd.read_csv(
    RUPAWS_DIR / "test/rupaws_qqp_test.csv",
    sep=";"
)

wiki_test = pd.read_csv(
    RUPAWS_DIR / "test/rupaws_wiki_test.csv",
    sep=";"
)
```

```python
qqp_test.head()
```

```python
wiki_test.head()
```

```python
wiki_test = wiki_test[["sentence1", "sentence2", "label"]]
```

```python
rupaws_test = pd.concat(
    [qqp_test, wiki_test],
    ignore_index=True
)

rupaws_test.shape
```

```python
rupaws_train["label"].value_counts().sort_index()
```

```python
rupaws_test["label"].value_counts().sort_index()
```

```python
print("QQP")
print(qqp_test["label"].value_counts().sort_index())

print("\nWIKI")
print(wiki_test["label"].value_counts().sort_index())
```

```python
train_full = pd.concat(
    [
        paraphraser_train_bin,
        rupaws_train[["sentence1", "sentence2", "label"]]
    ],
    ignore_index=True
)
```

```python
print(train_full.shape)
print(train_full["label"].value_counts())
print(train_full["label"].value_counts(normalize=True))
```

```python
train_full[["sentence1", "sentence2", "label"]].isna().sum()
```

```python
print("Полные дубликаты:", train_full.duplicated().sum())

print(
    "Дубликаты пар:",
    train_full[["sentence1", "sentence2"]].duplicated().sum()
)
```

```python
train_full = train_full.drop_duplicates(
    subset=["sentence1", "sentence2"]
).reset_index(drop=True)

print(train_full.shape)
```

```python
pairs = train_full[["sentence1", "sentence2"]].astype(str)

reverse_pairs = pd.DataFrame({
    "sentence1": pairs["sentence2"],
    "sentence2": pairs["sentence1"]
})

print(
    "Обратных пар:",
    train_full[["sentence1", "sentence2"]].merge(
        reverse_pairs,
        on=["sentence1", "sentence2"],
        how="inner"
    ).shape[0]
)
```

```python
train_full["pair_key"] = train_full.apply(
    lambda x: tuple(sorted([x["sentence1"], x["sentence2"]])),
    axis=1
)
```

```python
train_full = train_full.drop_duplicates(
    subset=["pair_key"]
).reset_index(drop=True)

train_full = train_full.drop(columns=["pair_key"])
```

```python
print("Размер:", train_full.shape)
print("Дубликаты:", train_full[["sentence1", "sentence2"]].duplicated().sum())
```

```python
print(train_full.shape)
print(train_full["label"].value_counts())
print(train_full["label"].value_counts(normalize=True))
```

## Splitting dataset

```python
from sklearn.model_selection import train_test_split

train_df, val_df = train_test_split(
    train_full,
    test_size=0.15,
    random_state=42,
    stratify=train_full["label"]
)

print("train:", train_df.shape)
print("val:", val_df.shape)

print("\nTrain:")
print(train_df["label"].value_counts())

print("\nValidation:")
print(val_df["label"].value_counts())
```

```python
from pathlib import Path

processed_dir = Path("processed")
processed_dir.mkdir(parents=True, exist_ok=True)

train_df.to_csv(
    processed_dir / "train.csv",
    index=False
)

val_df.to_csv(
    processed_dir / "val.csv",
    index=False
)

paraphraser_test_bin.to_csv(
    processed_dir / "test_paraphraser.csv",
    index=False
)

rupaws_test.to_csv(
    processed_dir / "test_rupaws.csv",
    index=False
)
```
